### Codigo C++

```python
#include <iostream>
    
#include <chrono>//Libreria manejo tiempo en segundos

#include <thread>//Libreria agrega delay

#include <string>// Libreria strings

#include <cstdlib>//Libreria generar numero random

using namespace std;
using namespace std::this_thread;// para no tener que escribir this_thread:: antes de sleep_for()
using namespace std::chrono; //para no tener que escribir chrono:: antes de seconds()
int main(){

    int resultado, eleccion, lanzamiento;
    string enter;
    lanzamiento=0;
	cout<< "========================="<<endl;
	cout<< "Bienvenido a Cara o Sello"<<endl;
	cout<< "========================="<<endl;
	cout<< "Presione Enter para seguir:";
    getline(cin, enter);//Lee el enter

	cout<< "¡Se ha lanzado una moneda!"<<endl;

    //Simulacion del lanzamiento
	for (int i=0;lanzamiento<2;i++){
    	cout<< "|"<<endl;
        sleep_for(seconds(1));//sleep_for agrega el delay y seconds agrega el tiempo del delay
		cout<< "/"<<endl;
        sleep_for(seconds(1));
		cout<< "-"<<endl;
        sleep_for(seconds(1));
		cout<<"\\"<<endl;
        sleep_for(seconds(1));
		lanzamiento +=1;
	}

	cout<<"¡La moneda ha caido!"<<endl;
	resultado = rand() % 2 + 1; //Generacion de un numero random del 1 al dos

	cout<<"Escribe tu opcion para revelar si has acertado"<<endl;
	cout<<"1. Sello"<<endl;
	cout<<"2. Cara"<<endl;
	cout<<"Mi eleccion es:";
	cin>>eleccion;

    //Imprime el lado que ha caido
	if (resultado==1){
	    cout<<"¡¡Ha caido Sello!!"<<endl;
	}
	if (resultado==2){
	    cout<<"¡¡Ha caido Cara!!"<<endl;
	}
    //Muestra al jugador si ha ganado o ha perdido.
	if (eleccion==resultado){
	    cout<<"¡Felicidades, has acertado!"<<endl;
	}
	else {
	    cout<<"Has fallado. Intenta de nuevo."<<endl;
	}
	return 0;
    
}
```
