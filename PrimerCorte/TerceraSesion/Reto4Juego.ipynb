### Codigo C++:

```python
#include <iostream>
#include <cstdlib> //Libreria que genera el numero 
#include <thread> // Libreria que agrega el delay
#include <chrono> //Libreria que agrega el tiempo del delay
#include <string> //string

using namespace std;
using namespace std::this_thread;
using namespace std::chrono;

int main() {
	int eleccion, resultado;
	string enter;

	cout << "====================================" << endl;
	cout << "Bienvenido a Piedra, Papel o Tijera" << endl;
	cout << "====================================" << endl;
	cout << "Presione Enter para seguir: ";
	getline(cin, enter);

	cout << "Seleccione la opcion a jugar (La CPU no hara trampa):" << endl;
	cout << "1. Piedra." << endl;
	cout << "2. Papel." << endl;
	cout << "3. Tijera." << endl;
	cout << "Opcion: ";
	cin >> eleccion;

	if (eleccion < 1 || eleccion > 3) {
		cout << "Seleccione una opcion correcta. Vuelva a Intentarlo." << endl;
	} else {
		// Generar número aleatorio entre 1 y 3
		resultado = rand() % 3 + 1;

        //Inicio juego
		cout << "¡¡¡¡Que inicie el juego!!!" << endl;
		sleep_for(seconds(1));
		cout << "Piedra" << endl;
		sleep_for(seconds(1));
		cout << "Papel" << endl;
		sleep_for(seconds(1));
		cout << "¡O Tijera!" << endl;
		sleep_for(seconds(1));
		cout << endl;

		// Mostrar la elección de la CPU
		if (resultado == 1) {
			cout << "La CPU eligio: Piedra" << endl;
		} else if (resultado == 2) {
			cout << "La CPU eligio: Papel" << endl;
		} else if (resultado == 3) {
			cout << "La CPU eligio: Tijera" << endl;
		}

		// Mostrar el ganador
		if (eleccion == resultado) {
			sleep_for(seconds(1));
			cout << "¡Es un Empate!" << endl;
		} else if ((resultado == 1 && eleccion== 2) || (resultado == 2 && eleccion== 3) ||(resultado == 3 && eleccion== 1)) {
            sleep_for(seconds(1));
			cout << "¡Ganaste! Felicitaciones." << endl;
		} else {
            sleep_for(seconds(1));
			cout << "¡Perdiste! La CPU gana esta vez." << endl;
		}
	}
	return 0;
}
```
