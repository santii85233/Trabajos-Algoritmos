Algoritmo RetoVasos

	Definir vaso_A, vaso_B, vaso_C Como Cadena
	
	// Estado Base
	vaso_A <- "Jugo de naranja"
	vaso_B <- "Agua"
	vaso_C <- "Vacio"
	
	Escribir "==== ESTADO BASE ===="
	Escribir "Vaso A tiene: ", vaso_A
	Escribir "Vaso B tiene: ", vaso_B
	Escribir "Vaso C tiene: ", vaso_C
	Escribir "========================"
	Escribir ""
	
	Escribir "PRIMER PASO"
	Escribir ">>Agarro el Vaso A y el Vaso C de la mesa."
	Escribir ">> Paso el Jugo de naranja del Vaso A a el Vaso C."
	vaso_C <- vaso_A     // El Vaso C ahora tiene lo que tenía el A
	vaso_A <- "Vacio"   // el Vaso A se quedo vacio
	Escribir ">> Dejo el Vaso A y el Vaso C en la mesa."
	Escribir "   (Estado actual -> Vaso A: ", vaso_A, " | Vaso C: ", vaso_C, ")"
	Escribir ""
	
	Escribir "SEGUNDO PASO"
	Escribir ">> Agarro el Vaso B y el Vaso A de la mesa."
	Escribir ">> paso el Agua del Vaso B a el Vaso A."
	vaso_A <- vaso_B     // el Vaso A tiene el agua
	vaso_B <- "Vacio"    // el Vaso B se quedo vacio
	Escribir ">> Dejo el Vaso B y el Vaso A en la mesa."
	Escribir "   (Estado actual -> Vaso B: ", vaso_B, " | Vaso A: ", vaso_A, ")"
	
	Escribir ""
	
	Escribir "TERCER PASO"
	Escribir ">> Agarro el Vaso C y el Vaso B de la mesa."
	Escribir ">> paso el Jugo del Vaso C a el Vaso B."
	vaso_B <- vaso_C     // el Vaso B recibe el Jugo de naranja
	vaso_C <- "Vacio"    //el Vaso C vuelve a estar vacio.
	Escribir ">> Dejo el Vaso C y el Vaso B en la mesa."
	Escribir "   (Estado actual -> Vaso C: ", vaso_C, " | Vaso B: ", vaso_B, ")"
	
	Escribir ""
	
	Escribir "==== ESTADO FINAL ===="
	Escribir "Vaso A tiene: ", vaso_A
	Escribir "Vaso B tiene: ", vaso_B
	Escribir "Vaso C tiene: ", vaso_C
	Escribir "========================"
FinAlgoritmo

### Lenguaje Natural:

**Estado Base**:

**En la mesa tenemos tres vasos:**

-El Vaso A tiene jugo de naranja.

-El Vaso B tiene agua.

-El Vaso C está vacío.

**Primer Paso**

-Agarro el Vaso A y el Vaso C de la mesa.

-Paso el jugo de naranja del Vaso A al Vaso C.

-Dejo el Vaso A y el Vaso C en la mesa.

(Estado actual: el Vaso A quedó vacío y el Vaso C tiene el jugo de naranja).

**Segundo Paso**

-Agarro el Vaso B y el Vaso A de la mesa.

-Paso el agua del Vaso B al Vaso A.

-Dejo el Vaso B y el Vaso A en la mesa.

(Estado actual: el Vaso B quedó vacío y el Vaso A tiene el agua).

**Tercer Paso**

-Agarro el Vaso C y el Vaso B de la mesa.

-Paso el jugo del Vaso C al Vaso B.

-Dejo el Vaso C y el Vaso B en la mesa.

(Estado actual: el Vaso C vuelve a quedar vacío y el Vaso B tiene el jugo de naranja).

**Estado Final**

Al terminar, el resultado en la mesa es:

-El Vaso A tiene agua.

-El Vaso B tiene jugo de naranja.

-El Vaso C está vacío.
