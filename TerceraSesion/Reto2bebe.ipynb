#include <iostream>

using namespace std;

int main(){

    float dosisVacuna, pesoBebe, mesesBebe;
    cout<< "Ingrese el peso del bebe en kg:";
    cin>>pesoBebe;
    cout<<"Ingrese los meses que tiene el bebe:";
    cin>>mesesBebe;
    dosisVacuna = ((pesoBebe+10)/(mesesBebe*10))*8;
    cout<<"La dosis que se le debe aplicar al bebe es de: "<< dosisVacuna <<"ml"<<endl;
    return 0;
    
}
