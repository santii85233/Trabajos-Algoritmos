### Pseudocodigo:

Algoritmo Adivinar

	Definir inferior, superior, intento, adivinanza, respuesta Como Entero
	Definir acertado Como Lógico
	Definir continuar Como Cadena
	// Iniciamos las variables del rango
	inferior <- 1
	superior <- 100
	intento <- 1
	acertado <- Falso
	
	Escribir 'Guessber: Software Adivinador'
	Escribir 'Piensa un numero del 1 al 100 en secreto.'
	Escribir 'Con mi estrategia, garantizo adivinarlo en 7 intentos o menos.'
	Escribir 'Presiona Enter una vez pensado.(Cambiarlo durante la adivinanza es trampa)'
	Leer continuar
	Mientras acertado=Falso Y intento<=7 Hacer
		// La estrategia es siempre adivinar el numero exactamente en medio del rango
		adivinanza <- trunc((inferior+superior)/2)
		Escribir ''
		Escribir 'Intento #', intento, ': ¿Tu numero es el ', adivinanza, '?'
		Escribir 'Responde ingresando el numero:'
		Escribir '1 -> No, mi numero es MAYOR'
		Escribir '2 -> No, mi numero es MENOR'
		Escribir '3 -> ¡CORRECTO!'
		Leer respuesta
		Si respuesta=1 Entonces
			// si el numero es mayor, descartamos toda la mitad inferior del rango
			inferior <- adivinanza+1
			intento <- intento+1
		SiNo
			Si respuesta=2 Entonces
				// el numero es menor, descartamos toda la mitad superior del rango
				superior <- adivinanza-1
				intento <- intento+1
			SiNo
				Si respuesta=3 Entonces
					acertado <- Verdadero
					Escribir '¡Adivine tu numero en el intento ', intento, '!'
				SiNo
					Escribir 'Opcion no valida. No se contabilizara este intento.'
				FinSi
			FinSi
		FinSi
	FinMientras
	// si se acaban los intentos y no se adivino, es por un error del usuario
	Si acertado=Falso Entonces
		Escribir ''
		Escribir 'Se acabaron los 7 intentos sin exito.'
		Escribir 'Es posible que hayas dado una pista erronea.'
		Escribir '!NO ENGAÑAS A GUESSBER!'
	FinSi
FinAlgoritmo

### Lenguaje Natural:

**En el sistema de adivinanza tenemos los rangos iniciales y el control del juego:**

-Un límite inferior inicializado en 1.

-Un límite superior inicializado en 100.

-Un contador de intentos que inicia en 1 (con un máximo de 7)

-Una variable de respuesta del usuario y el estado de adivinanza (acertado o no)

**Primer Paso**

-Inicio la interfaz del software "Guessber".

-Solicito al usuario que piense un número en secreto del 1 al 100

-Espero a que el usuario presione Enter para continuar

**Segundo Paso**

-Calculo la adivinanza seleccionando el número exacto a la mitad del rango actual

-Muestro la adivinanza y solicito la respuesta del usuario (1 para MAYOR, 2 para MENOR, 3 para CORRECTO)

-Si la respuesta es "CORRECTO", marco la adivinanza como acertada y muestro el mensaje de victoria


**Tercer Paso**

-Si la respuesta es "MAYOR", descarto la mitad inferior ajustando el límite inferior a la adivinanza más 1 y aumento el intento.

-Si la respuesta es "MENOR", descarto la mitad superior ajustando el límite superior a la adivinanza menos 1 y aumento el intento.

-Si se ingresa una opción no válida, muestro un error sin gastar un intento.

-Repito el proceso (Paso 2 y 3) mientras no se haya acertado y no se superen los 7 intentos.


**Al terminar, el resultado del sistema es:**

-La confirmación de haber adivinado el número indicando la cantidad de intentos utilizados.

-O si se agotaron los 7 intentos sin adivinar, un mensaje notificando que el usuario dio información incorrecta.
