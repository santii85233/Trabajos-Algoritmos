		tiene_urgencia <- Verdadero
	SiNo
		tiene_urgencia <- Falso
	FinSi

		// La urgencia vital no importa edad ni tiempo, pasa con prioridad alta

		Si edad < 18 O edad >= 65 Entonces

### Lenguaje Natural:

**En el sistema de clasificación tenemos tres datos de entrada:**

-La edad del paciente.
-Si el paciente tiene una condición de urgencia vital.
-Los minutos que lleva esperando en la fila.

**Primer Paso**

-Ingreso la edad del paciente.

-Ingreso si tiene o no urgencia vital la cual esta determinada por unos dolores tipicos de una urgencia (1 para Si, 2 para No).
-Ingreso el tiempo de espera en minutos.

**Segundo Paso**

-Evaluo si el paciente presenta una urgencia vital.

-Si la respuesta es si, se le da una prioridad "ALTA".

-Si no, se procede a evaluar su edad y su tiempo de espera.

**Tercer Paso**

-Si es menor de 18 o mayor/igual a 65 años y esperó más de 60 minutos, se le asigna prioridad "ALTA"; si esperó menos, prioridad "MEDIA".

-Si es un adulto (18 a 64 años) y esperó más de 120 minutos, se le asigna prioridad "MEDIA"; si esperó menos, prioridad "BAJA".

**Estado Final**

Al terminar, el resultado del sistema es:

-La prioridad asignada al paciente ("ALTA", "MEDIA" o "BAJA").
